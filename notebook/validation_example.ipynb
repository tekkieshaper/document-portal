```python
from pydantic import BaseModel, RootModel
```

```python
class ChangeFormat(BaseModel):
    pageno: int
    lineno: int
    changes:str
```

```python
class SummaryResponse(RootModel[list[ChangeFormat]]):
    pass
```

```python
change1= ChangeFormat(pageno=1, lineno=1, changes="test 1")
change2= ChangeFormat(pageno=2, lineno=2, changes="test 2")
```

```python
response = SummaryResponse([change1,change2])
```

```python
print(response)
```
