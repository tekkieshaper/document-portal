```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(model="gemini-embedding-exp-03-07")
```

```python
embedding_model.embed_query("hello world !")
```

```python
from langchain.document_loaders import PyPDFLoader
```

```python
folder_path = os.path.join(os.getcwd(),"data")
```

```python
folder_path
```

```python
all_documents=[]
```

```python
os.listdir(folder_path)
```

```python
#loop through the pdf files in the folder
for filename in os.listdir(folder_path):
    if filename.endswith(".pdf"):
        file_path = os.path.join(os.getcwd(),"data",filename)
        loader = PyPDFLoader(file_path)
        documents = loader.load()
        all_documents.extend(documents)
```

```python
all_documents
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap = 5,
    length_function = len
)
```

```python
documents = text_splitter.split_documents(all_documents)
```

```python
documents
```

```python
from langchain.vectorstores import FAISS
```

```python
vectorstore = FAISS.from_documents(documents,embedding_model)
```

```python
vectorstore
```

```python
#how much sleep reduced due to street level noise for the subjects
vectorstore.similarity_search("what is the API request body?",k=3)
```

```python
retriever = vectorstore.as_retriever()
```

```python
retriever.invoke("how much sleep reduced due to street level noise for the subjects")
```

```python
prompt_template ="""
    Answer the question based on the context provided below.  
    If the context doesn't has sufficient information then respond
    "I don't have enough information"
    Context :{context}
    Question : {question}
    Answer : 
"""
```

```python
from langchain.prompts import PromptTemplate
```

```python
prompt = PromptTemplate(template= prompt_template,
                        input_variables=["context","question"])
```

```python
prompt
```

```python
from langchain_core.output_parsers import StrOutputParser 
```

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
doc_content = retriever | format_docs
doc_content
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
rag_chain = (
    {"context" :retriever |format_docs , "question":RunnablePassthrough()}
    | prompt
    | llm_model
    | StrOutputParser()
)
```

```python
#what is the planned next step to get precise sleep pattern
response = rag_chain.invoke("what is the integration modules rate?")
```
