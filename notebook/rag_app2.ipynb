```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="qwen/qwen3-32b")
```

```python
llm.invoke("what is capital of India?")
```

```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
```

```python
embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001")
```

```python
embeddings.embed_query("hello world !")
```

```python
from langchain.document_loaders import PyPDFLoader
```

```python
pdffilepath = os.path.join(os.getcwd(),"data","API Documentation Snippet.pdf")
```

```python
pdfload = PyPDFLoader(pdffilepath)
```

```python
document = pdfload.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 5,
     length_function = len
)
```

```python
docs = splitter.split_documents(document)
```

```python
docs
```

```python
from langchain.vectorstores import FAISS
```

```python
vectorstore = FAISS.from_documents(docs,embeddings)
```

```python
vectorstore.similarity_search("what is the email ID in the end point examle ?")
```

```python
retriever = vectorstore.as_retriever()
```

```python
retriever.invoke("what is the rate limiter applied in api end point /user/login ?")
```

```python
from langchain.prompts import PromptTemplate
```

```python
def format_doc(docs):
    return "\n\n".join([doc.page_content  for doc in docs])
```

```python
# prompt_template ="""
#   Answer the question with following questions.
#   If you don't have answer then reply that you don't have enough information 

#   Context : {context}
#   Question : {question}
#   Answer : 
# """
prompt_template ="""
    Answer the question based on the context provided below.  
    If the context doesn't has sufficient information then respond
    "I don't have enough information"
    Context :{context}
    Question : {question}
    Answer : 
"""
```

```python
prompt = PromptTemplate(
    template= prompt_template,
    input_variables=["context","question"]
)
```

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
doc_content = retriever | format_doc
```

```python
doc_content
```

```python
rag_chain = (
     {"context": retriever | format_doc,"question": RunnablePassthrough()}
    | prompt
    | llm
    | StrOutputParser()
)
```

```python
rag_chain
```

```python
rag_chain.invoke("what is the rate limiter of the end point ?")
```
